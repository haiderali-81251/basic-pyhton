# 1) Area of a Rectangle

Write a Python program to calculate the area of a rectangle.

**Formula:**

$$
\text{Area} = \text{length} \times \text{width}
$$

Take **length** and **width** as inputs from the user.

```python
length=float(input("Enter length of an rectangle :"))
width=float(input("Enter width of an rectangle :"))

area=length*width

print(f"Area of rectangle if {area}")
```

# 3) Simple Interest

Write a Python program to calculate the simple interest.

**Formula:**

$$
\text{Simple Interest} = \text{Principal} \times \text{Rate} \times \text{Time}
$$

Take **Principal**, **Rate**, and **Time** as inputs from the user.

```python
principal=int(input("Enter principle amount $ :"))
rate=float(input("Enter interest rate per year % :"))
time=float(input("Enter time period in years:"))

simple_interest=principle*(rate/100)*time

print(f"simple interest amount will be {simple_interest:.2f}$")
```

# 7) Compound Interest

Write a Python program to calculate compound interest.

**Formula:**

$$
A = P\left(1 + \frac{r}{n}\right)^{nt}
$$

**Where:**

- **A** = Total amount
- **P** = Principal amount
- **r** = Annual interest rate (decimal)
- **n** = Number of times interest is compounded per year
- **t** = Time in years

Take **P**, **r**, **n**, and **t** as inputs from the user.

```python
principle_amount=int(input("Enter principle amount $ :"))
annual_interest_rate=float(input("Enter annual interest rate :"))
no_of_compounding=int(input("no of time interest compounded in year"))
time=float(input("time in years :"))

total_amount=principle_amount * (1+annual_interest_rate/no_of_compounding)**(no_of_compounding*time)

print(f"the total amount will be {total_amount:.2f}$")
```

# 13) Gravitational Force

Write a Python program to calculate the gravitational force between two objects.

**Formula:**

$$
F = G\frac{m_1 \times m_2}{r^2}
$$

**Where:**

- **m₁, m₂** = masses of the objects
- **r** = distance between the centers of the objects
- **G** = gravitational constant

Take **m₁**, **m₂**, and **r** as inputs from the user.

```python
mass_1=int(input("Enter mass of object 1 :"))
mass_2=int(input("Enter mass of object 2 :"))
distance=int(input("Enter distance between objects :"))

force=((6.67*10**-11)*mass_1*mass_2)/distance**2

print(f"the gravitational force between two objects is {force} Newtons")
```

Write a Python program to calculate the future value of an investment.

$$
FV = PV(1+r)^t
$$

- **FV** = Future value
- **PV** = Present value
- **r** = Annual interest rate (as a decimal)
- **t** = Time in years

Take **PV**, **r**, and **t** as inputs from the user.

# 19) Work Done by a Force

Write a Python program to calculate the work done by a force.

**Formula:**

$$
W = Fd\cos(\theta)
$$

**Where:**

- **W** = Work done
- **F** = Force
- **d** = Distance
- **θ** = Angle between force and direction of movement (in degrees)

Take **F**, **d**, and **θ** as inputs from the user.

```python
import math

force=int(input("Enter force applied :"))
distance=int(input("Enter distance travelled :"))
angle=int(input("Enter angle between force and direction of motion"))

work_done=force*distance*(math.cos(math.radians(angle)))

print(f"the work done is {work_done:.2f}")
```

# 20) Heat Transfer

Write a Python program to calculate the amount of heat transferred.

**Formula:**

$$
Q = mc\Delta T
$$

**Where:**

- **Q** = Heat transfer
- **m** = Mass
- **c** = Specific heat capacity
- **ΔT** = Change in temperature

Take **m**, **c**, and **ΔT** as inputs from the user.

```python
mass=float(input("Enter mass of object :"))
specific_heat_capacity=float(input("Enter specific heat capacity of object :"))
temperature_change=float(input("Enter change in temperature :"))

heat_transfer=mass*specific_heat_capacity*temperature_change

print(f"The heat transfered to/from object is {heat_transfer}")

```
